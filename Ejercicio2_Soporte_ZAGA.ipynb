# Ejercicio 2: Escalamiento de tickets de soporte técnico

Este ejercicio desarrolla un agente inteligente que decide a qué nivel de soporte enviar un ticket considerando el tiempo de espera, el nivel de urgencia y si el cliente es premium.

### Justificación de las reglas

Los tickets de urgencia alta deben escalarse inmediatamente para reducir el tiempo de atención de problemas críticos. En los tickets de urgencia media, el tiempo de espera permite determinar cuándo es necesario escalar. La condición de cliente premium puede aumentar la prioridad cuando la espera ya es considerable. Estas reglas buscan equilibrar una atención oportuna al cliente con el uso adecuado de los recursos del equipo de soporte.


```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):

    nivel_urgencia = nivel_urgencia.lower()

    if nivel_urgencia not in ["baja", "media", "alta"]:
        return "rechazar ticket", "nivel de urgencia no válido"

    # Regla 1:
    # Una urgencia alta se escala inmediatamente.
    if nivel_urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta reportada por el cliente"

    # Regla 2:
    # Urgencia media con espera de 30 minutos o más.
    if nivel_urgencia == "media" and tiempo_espera_minutos >= 30:
        return "escalar a nivel 2", "urgencia media con espera de 30 minutos o más"

    # Regla 3:
    # Cliente premium con urgencia media y espera de 20 minutos o más.
    if nivel_urgencia == "media" and cliente_premium and tiempo_espera_minutos >= 20:
        return "escalar a nivel 2", "cliente premium con urgencia media y espera considerable"

    # Regla 4:
    # Cliente premium con urgencia baja y espera de 60 minutos o más.
    if nivel_urgencia == "baja" and cliente_premium and tiempo_espera_minutos >= 60:
        return "escalar a nivel 2", "cliente premium con urgencia baja y espera de 60 minutos o más"

    # Regla 5:
    # Cualquier ticket de baja urgencia con espera de 90 minutos o más.
    if nivel_urgencia == "baja" and tiempo_espera_minutos >= 90:
        return "escalar a nivel 2", "ticket de baja urgencia con espera prolongada"

    # Regla 6:
    # Si ninguna condición anterior se cumple.
    return "mantener en nivel 1", "las condiciones no requieren escalamiento inmediato"
```

### Simulación y pruebas

Se realizan diferentes pruebas combinando el tiempo de espera, el nivel de urgencia y la condición de cliente premium. Se incluyen casos en los que las condiciones no coinciden, como un cliente premium con urgencia baja pero con un tiempo de espera elevado.


```python
casos_soporte = [
    (5, "alta", False),
    (10, "media", False),
    (30, "media", False),
    (20, "media", True),
    (60, "baja", True),
    (90, "baja", False),
    (40, "baja", True),
    (100, "baja", True)
]

for espera, urgencia, premium in casos_soporte:

    accion, motivo = agente_soporte(
        espera,
        urgencia,
        premium
    )

    print("Tiempo de espera:", espera, "minutos")
    print("Nivel de urgencia:", urgencia)
    print("Cliente premium:", premium)
    print("Decisión:", accion)
    print("Motivo:", motivo)
    print("-" * 60)
```
