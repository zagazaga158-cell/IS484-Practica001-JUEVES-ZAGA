# Ejercicio 1: Evaluación de solicitud de crédito bancario

Este ejercicio desarrolla un agente inteligente que evalúa automáticamente una solicitud de crédito bancario considerando el ingreso mensual, el monto solicitado y el historial moroso del solicitante.


### Ficha PEAS

| Elemento                    | Descripción                                                                                                                                     |
| --------------------------- | ----------------------------------------------------------------------------------------------------------------------------------------------- |
| **P – Medida de desempeño** | Tomar decisiones coherentes, reducir el riesgo crediticio y proporcionar un motivo claro para cada decisión.                                    |
| **E – Entorno**             | Banco, clientes, solicitudes de crédito y sistema de evaluación crediticia.                                                                     |
| **A – Acciones**            | Aprobar, aprobar con condiciones o rechazar la solicitud.                                                                                       |
| **S – Percepción**          | Ingreso mensual, monto solicitado y existencia de historial moroso.                                                                             |
| **Objetivo**                | Evaluar automáticamente una solicitud de crédito utilizando la relación entre monto solicitado e ingreso mensual junto con el historial moroso. |


### Justificación de las reglas

La relación entre el monto solicitado y el ingreso mensual permite evaluar qué tan grande es el crédito solicitado respecto de los ingresos del cliente. Un historial moroso representa un mayor riesgo para el banco, por lo que se aplican condiciones más estrictas. Las solicitudes con una relación baja y sin historial moroso pueden aprobarse, mientras que relaciones mayores pueden requerir condiciones o rechazo. Estas reglas permiten tomar decisiones de forma automática y explicable.


```python
def agente_credito(ingreso_mensual, monto_solicitado, tiene_historial_moroso):
    relacion = monto_solicitado / ingreso_mensual

    if tiene_historial_moroso and relacion > 0.5:
        return "rechazar", "relacion monto/ingreso mayor a 0.5 con historial moroso"

    if relacion > 1.0:
        return "rechazar", "relacion monto/ingreso mayor a 1.0"

    if tiene_historial_moroso:
        return "aprobar con condiciones", "historial moroso detectado"

    if relacion > 0.5:
        return "aprobar con condiciones", "relacion monto/ingreso entre 0.5 y 1.0"

    return "aprobar", "relacion monto/ingreso igual o menor a 0.5 y sin historial moroso"
```

```python
ingreso = 3000
monto = 1500

relacion = monto / ingreso

print(relacion)
```

```python
casos_credito = [
    (3000, 1000, False),
    (3000, 1500, False),   # Caso límite: relación = 0.5
    (3000, 1800, False),
    (3000, 1500, True),
    (3000, 1800, True),
    (3000, 3600, False)
]

for ingreso, monto, moroso in casos_credito:
    accion, motivo = agente_credito(
        ingreso,
        monto,
        moroso
    )

    relacion = monto / ingreso

    print("Ingreso mensual:", ingreso)
    print("Monto solicitado:", monto)
    print("Historial moroso:", moroso)
    print("Relación:", round(relacion, 2))
    print("Decisión:", accion)
    print("Motivo:", motivo)
    print("-" * 60)
```

### Visualización

Se generan 200 solicitudes de crédito aleatorias variando el ingreso mensual, el monto solicitado y el historial moroso. Luego se calcula la decisión de cada solicitud y se representa gráficamente la relación monto/ingreso frente al ingreso mensual.


```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 200

ingresos = np.random.randint(1000, 8001, n)
montos = np.random.randint(300, 10001, n)
historiales = np.random.choice([False, True], size=n)

relaciones = montos / ingresos

decisiones = []

for ingreso, monto, moroso in zip(
    ingresos,
    montos,
    historiales
):
    accion, motivo = agente_credito(
        ingreso,
        monto,
        moroso
    )

    decisiones.append(accion)

acciones = [
    "aprobar",
    "aprobar con condiciones",
    "rechazar"
]

plt.figure(figsize=(10, 6))

for accion in acciones:

    indices = [
        i for i, decision in enumerate(decisiones)
        if decision == accion
    ]

    plt.scatter(
        relaciones[indices],
        ingresos[indices],
        label=accion,
        alpha=0.7
    )

plt.xlabel("Relación monto solicitado / ingreso mensual")
plt.ylabel("Ingreso mensual (S/)")
plt.title("Evaluación de 200 solicitudes de crédito")

plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
